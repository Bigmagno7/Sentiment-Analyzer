```python
from transformers import pipeline
import gradio as gr

sentiment_pipeline = pipeline("sentiment-analysis")

def analizza_sentimento(frase):
    risultato = sentiment_pipeline(frase)[0]
    label = risultato['label']
    score = round(risultato['score'], 3)

    if label == "NEGATIVE":
        emoji = "😠"
    elif label == "POSITIVE":
        emoji = "😊"
    else:
        emoji = "😐"

    return f"{emoji} {label} ({score})"

interfaccia = gr.Interface(
    fn=analizza_sentimento,
    inputs=gr.Textbox(label="Inserisci una frase"),
    outputs=gr.Textbox(label="Risultato"),
    title="Analizzatore di Sentimento",
    description="Scrivi una frase e scopri se è positiva o negativa!"
)

interfaccia.launch()
```
